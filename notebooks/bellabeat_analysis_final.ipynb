## Activity and Sleep Relationship

Daily activity and sleep records were matched by user and recorded date. For this case study, `SleepDay` is interpreted as the date the sleep period begins, so the same-date merge compares activity during that date with sleep beginning that night.

Because users contributed different numbers of matched records, this is treated as a **pooled exploratory user-day analysis**. Correlation is used only to describe association; it does not imply causation.

```python
activity_sleep_same_day = sleep_daily.merge(
    activity_daily,
    left_on=["Id", "SleepDay"],
    right_on=["Id", "ActivityDate"],
    how="inner"
)

print("Matched same-day records:", activity_sleep_same_day.shape[0])
```

```python
same_day_correlation = activity_sleep_same_day["TotalMinutesAsleep"].corr(
    activity_sleep_same_day["TotalSteps"]
)
round(same_day_correlation, 2)
```

```python
x = activity_sleep_same_day["TotalSteps"]
y = activity_sleep_same_day["TotalMinutesAsleep"] / 60

fig, ax = plt.subplots(figsize=(8, 5))

ax.scatter(x, y, alpha=0.45, s=30)

m, b = np.polyfit(x, y, 1)
x_line = np.linspace(x.min(), x.max(), 100)
ax.plot(x_line, m * x_line + b, color="red", linewidth=2)

ax.set_title(
    "Daily Steps Show Little Relationship With Sleep Duration",
    fontsize=14,
    fontweight="bold",
    pad=28
)

ax.text(
    0.5,
    1.02,
    f"Pearson correlation: r = {same_day_correlation:.2f} | "
    f"{len(activity_sleep_same_day)} matched user-days",
    transform=ax.transAxes,
    ha="center",
    fontsize=10
)

ax.set_xlabel("Daily Steps")
ax.set_ylabel("Sleep Duration (Hours)")
ax.grid(axis="y", alpha=0.15)

fig.tight_layout()

fig.savefig(
    PROJECT_ROOT / "images" / "steps_vs_sleep.png",
    dpi=150,
    bbox_inches="tight"
)

plt.show()
```

### Relationship Between Activity and Sleep

- Total steps during the day and sleep duration beginning that night showed a very weak negative correlation (`r ≈ -0.19`).
- The scatter plot showed substantial variation and no strong linear relationship between daily step count and sleep duration.

Overall, the data does not provide evidence of a strong relationship between daily step count and sleep duration. This result is exploratory, reflects pooled matched user-days, and should not be interpreted as causal.

## Key Findings

### 1. Activity Patterns

Weekday activity was relatively consistent, at roughly 7,000 - 8,000 average daily steps. Average activity was slightly higher on Tuesdays and Saturdays with over 8,000 average daily steps. Sunday saw a lower daily average at roughly 6,900 steps.

Despite the relatively stable weekday averages, there was substantial variation between users: 25% of users averaged below roughly 5,500 daily steps, while the top 25% averaged above roughly 9,500 daily steps.

### 2. Sleep Patterns

Average sleep duration was relatively consistent across the week, with only modest differences between weekdays. Sunday recorded the highest average sleep duration at 7.55 hours, while Thursday recorded the lowest at 6.69 hours. Sleep-data coverage varied considerably between users, so these weekday averages should be interpreted cautiously.

### 3. Steps vs Sleep Relationship

Upon analysis, there was a very weak negative correlation between daily steps and daily sleep duration. This means we tended to see slightly lower sleep durations on days with higher activity.

However, the correlation was very weak (r = -0.19), and the observations showed substantial variation. Therefore, the data does not provide strong evidence of a meaningful relationship between daily steps and sleep duration.

## Recommendations

### 1. Personalised Activity Goals

Daily activity was relatively consistent across weekdays, but activity levels varied substantially between individual users. Bellabeat could use each user's recent activity history to set personalised step goals and progress benchmarks rather than relying on one generic target.

This would make goals more relevant to each user's baseline and allow targets to adapt as behaviour changes over time. Marketing could position Bellabeat Time as a wellness watch that adapts to the user's own activity baseline rather than promoting one target for everyone.

### 2. Encourage Consistent Sleep Tracking

Sleep-data coverage varied substantially between users. If Bellabeat sees similar gaps in its own current user data, it could encourage regular sleep tracking through reminders and educational messaging explaining how more complete data enables more meaningful personalised insights and trends over time.

This could include in-app or wearable reminders when sleep data has not been recorded consistently. Marketing could also communicate that consistent overnight use helps Bellabeat Time provide more useful personal sleep trends.

### 3. Keep Activity and Sleep Messaging Separate

The weak relationship between daily steps and sleep duration (`r = -0.19`) suggests Bellabeat should avoid marketing messages that imply a simple connection between being more active and sleeping longer. Instead, activity and sleep should be treated as separate wellness behaviours with distinct goals, insights, and recommendations.


## Limitations

1. **Small sample size:** activity data covers 33 users and sleep data covers 24 users.
2. **Short observation period:** the main analysis covers roughly one month.
3. **Incomplete and questionable records:** user coverage varied considerably, particularly for sleep, and some unusual observations could not be confidently classified as errors.
4. **Limited demographic information:** age, gender, location and other demographic characteristics are unknown, limiting how confidently the findings can be generalised to Bellabeat's customer base.
5. **Unequal user contribution:** users contributed different numbers of recorded days, particularly for sleep, so the pooled activity-sleep correlation gives more weight to users with more matched records.
6. **Historical dataset:** the Fitbit data was collected in 2016. In a real business setting, I would want current user data before using these findings to guide marketing decisions in 2026.


## Conclusion

The analysis found that average activity and sleep patterns were relatively stable across the week, while individual activity levels varied widely and sleep-data coverage differed considerably between users.

Daily steps and sleep duration showed only a very weak relationship. For Bellabeat, the clearest opportunities are to personalise activity goals, encourage more consistent sleep tracking where similar gaps are present in current Bellabeat data, and keep activity and sleep messaging separate rather than implying a simple relationship between the two.

The findings are exploratory and should be interpreted within the limitations of the small sample, short observation period, uneven user coverage, limited demographic information and historical dataset.
